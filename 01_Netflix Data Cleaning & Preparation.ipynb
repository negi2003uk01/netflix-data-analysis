# Step 2: Identify and handle missing values

```python
# Identify: check how many missing values each column has
netflix.isnull().sum()
```

```python
netflix['country'] = netflix['country'].fillna('Unknown')   # fillna = fill missing values with something
netflix['cast'] = netflix['cast'].fillna('Unknown')
```

```python
netflix['rating'] = netflix['rating'].fillna(netflix['rating'].mode()[0])  # mode = most frequent value
```

```python
#  If a column is mostly empty and not useful, drop it
netflix = netflix.drop(columns=['director'])                 # drop = remove
```

```python
#remove rows that have missing values in critical columns  
#since dates are unique events, not repeating categories mode not work on this  
netflix = netflix.dropna(subset=['date_added', 'duration'])
```

# Step 3: Remove duplicate records and formatting inconsistencies

```python
# Duplicate = an exact repeated row
netflix = netflix.drop_duplicates()

```

```python
# Formatting inconsistency = same value written in different styles (e.g., "usa", "USA", "U.S.A")
netflix['country'] = netflix['country'].str.strip()        # strip = remove extra spaces
netflix['country'] = netflix['country'].str.title()         # title case = "United States" (capitalizes each word)
```

# Step 4: Standardize columns such as Country, Rating, and Type
