```python
import pandas as pd
import matplotlib.pyplot as plt   # library for creating charts/graphs
import seaborn as sns              # library built on matplotlib, makes prettier charts
custom_palette =['#E50914', '#221f1f', '#141414', '#B20710']

```

 # Step 1: Load the cleaned dataset

```python
# Set the custom palette
sns.set_palette(custom_palette)
netflix = pd.read_csv("netflix_cleaned.csv")   # load = read the file into memory
netflix.head()
```

# Step 2: Calculate the total number of Movies and TV Shows

```python
content_counts = netflix['type'].value_counts()   # counts how many times each unique value appears
print(content_counts)
```

 # Step 3: Create visualizations for content distribution

```python
# Bar chart (bar graph = rectangles showing comparison between categories)
plt.figure(figsize=(6,4))
sns.countplot(data= netflix,  x='type' , palette= ['#E50914', '#221f1f'] )
plt.title("Movies vs TV Shows on Netflix" , color= 'red')
plt.xlabel("Content Type" , color='red')
plt.ylabel("Count", color ='red')
plt.tight_layout()
plt.show()

```

```python
# Pie chart (circular chart = shows proportion/percentage of a whole)
plt.figure(figsize=(6,6))
plt.pie(content_counts, autopct='%1.1f%%', colors=['#e50914', '#221f1f']  ,wedgeprops={'edgecolor': 'black', 'linewidth': 2})
plt.title("Content Type Distribution" , color = 'red')
plt.ylabel("")
plt.tight_layout()
plt.show()
```

 # Netflix content primarily consists of 69.7% movies and 30.3% TV shows.It is evident that there are more Movies on Netflix than TV shows.

```python
rating_type = netflix.groupby(['type', 'rating']).size().reset_index(name='count')

plt.figure(figsize=(12,6))
# hue =adds a third dimension (extra grouping) by splitting bars into colors — here, splitting each country into Movie vs TV Show.
sns.barplot(data=rating_type, x='rating', y='count', hue='type')
plt.title("Content Rating Distribution: Movies vs TV Shows", color= 'red' )
plt.xlabel("Content Rating", color= 'red')
plt.ylabel("Count" , color = 'red')
plt.xticks(rotation=45)
plt.show()
```

# Step 4: Compare content proportions

```python
total = content_counts.sum()   # sum = total of all values
proportions = (content_counts / total) * 100   # proportion = a part compared to the whole, shown as %

print(proportions.round(2))   # round(2) = round the number to 2 decimal places
```

```python
movie_pct = proportions.get('Movie', 0)
tvshow_pct = proportions.get('TV Show', 0)

print(f"Movies make up {movie_pct:.1f}% of Netflix content.")
print(f"TV Shows make up {tvshow_pct:.1f}% of Netflix content.")
```

# Step 5: Summarize key findings

1) The Netflix dataset contains both Movies and TV Shows.
2) Movies make up the majority of the available content.
3) Approximately 69.7% of the titles are Movies, while 30.3% are TV Shows.
4) The visualizations clearly show that Netflix's library is more focused on Movies than TV Shows.
5) This suggests Netflix offers a wider selection of films, although TV Shows still represent a significant portion of the catalog.
